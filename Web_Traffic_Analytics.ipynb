# Web Traffic Analytics
Complete analysis of sessions, engagement, traffic sources, page paths and conversion funnel.

```python
import pandas as pd
df=pd.read_csv('data/web_traffic_sessions.csv',parse_dates=['date'])
df.head()
```

```python
pd.Series({'Sessions':len(df),'Bounce rate':df.bounce.mean(),'Engagement rate':df.engaged.mean(),'Avg pages/session':df.pages_viewed.mean(),'Avg duration (sec)':df.session_duration_sec.mean(),'Signup conversion':df.signup_complete.mean()})
```

```python
df.groupby('channel').agg(sessions=('session_id','count'),bounce_rate=('bounce','mean'),conversion=('signup_complete','mean')).sort_values('sessions',ascending=False)
```

```python
df.groupby('landing_page').agg(sessions=('session_id','count'),bounce_rate=('bounce','mean'),avg_pages=('pages_viewed','mean'),avg_duration=('session_duration_sec','mean')).sort_values('sessions',ascending=False)
```

```python
s=pd.Series({'Sessions':len(df),'Product View':df.product_view.sum(),'Pricing View':df.pricing_view.sum(),'Signup Started':df.signup_start.sum(),'Signup Completed':df.signup_complete.sum()})
pd.DataFrame({'users':s,'dropoff':1-s/s.shift(1)})
```
